# PYKT处理数据流程

```python
read_file = 'bridge_to_algebra_2006_2007_train.txt' 
```

```python
df = pd.read_table(read_file, low_memory=False)
df.columns, df.shape
```

```python
def replace_text(text):
    text = text.replace("_", "####").replace(",", "@@@@")
    return text



#concat problem name & step_name
df["Problem Name"] = df["Problem Name"].apply(replace_text)
df["Step Name"] = df["Step Name"].apply(replace_text)
df["Questions"] = df.apply(lambda x:f"{x['Problem Name']}----{x['Step Name']}",axis=1)
```

### 删除空值

```python
from datetime import datetime
def change2timestamp(t, hasf=True):
    if hasf:
        timeStamp = datetime.strptime(t, "%Y-%m-%d %H:%M:%S.%f").timestamp() * 1000
    else:
        timeStamp = datetime.strptime(t, "%Y-%m-%d %H:%M:%S").timestamp() * 1000
    return int(timeStamp)
```

```python
df = df.dropna(subset=["Anon Student Id", "KC(SubSkills)", "Questions", "Correct First Attempt", "First Transaction Time"])
#filter invalid record
df = df[df["Correct First Attempt"].isin([0,1])]
df.loc[:, "First Transaction Time"] = df.loc[:, "First Transaction Time"].apply(lambda t: change2timestamp(t))
df["KC(SubSkills)"] = df["KC(SubSkills)"].apply(replace_text)
df.shape
```

```python
df.columns
```

### 删除记录数小于3的

```python
# 统计每个 user_id 出现的次数
user_counts = df['Anon Student Id'].value_counts()
# 保留记录数大于或等于3的 user_id
filtered_df = df[df['Anon Student Id'].isin(user_counts[user_counts >= 3].index)]
df = filtered_df
```

重新选择特征"Anon Student Id", "KC(SubSkills)", "Questions", "Correct First Attempt", "First Transaction Time"

```python
# 初始化列分类
numerical_features = [
    'First Transaction Time',  # 第一次操作时间
   
]

categorical_features = [
    'Anon Student Id',  # 学生匿名ID
    'KC(SubSkills)',
    'Questions' , # 问题相关信息
    'Hints',  # 请求提示的次数
]

# 标签特征
label_feature = 'correct'  # 标签（即是否正确）
```

```python
# 选择需要保留的列
columns_to_keep = ['First Transaction Time', 'Anon Student Id', 'KC(SubSkills)', 'Questions' , 'Hints', 'correct']

# # 筛选保留指定列
df = df[columns_to_keep]
df.columns
```

归一化，去除极端值

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
import numpy as np

# 将数值特征转换为数值类型
for col in numerical_features:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 对数值特征进行归一化
scaler = MinMaxScaler()
df[numerical_features] = scaler.fit_transform(df[numerical_features])

# 观察归一化后的分布
for col in numerical_features:
    print(f"{col} max:", df[col].max(), "min:", df[col].min())

# 去除极端值，将值限制在 [0, 1] 范围内
for col in numerical_features:
    df[col] = np.clip(df[col], 0, 1)

# 处理类别特征：将其转换为整数编码
integer_encoded_features = {}
for col in categorical_features:
    df[col] = df[col].fillna('Unknown').astype(str)
    label_encoder = LabelEncoder()
    df[col] = label_encoder.fit_transform(df[col])
    integer_encoded_features[col] = {
        'encoder': label_encoder,
        'n_unique': len(label_encoder.classes_)
    }

# 处理标签特征：转换为整数编码
df[label_feature] = df[label_feature].fillna('Unknown').astype(str)
label_encoder = LabelEncoder()
df[label_feature] = label_encoder.fit_transform(df[label_feature])
```
